-i. Sequence to Sequence (Seq2Seq) Models

AI chatbots like ChatGPT don't just look at single words; they look at chains of words. We call them Sequence to Sequence models because they take a specific input chain and calculate the best output chain to match it.
A screenshot showing someone asking ChatGPT how it is today.

Think of it like a Language Translator. For example, a translator can't just swap words one-by-one; they have to read the whole sentence in English (the Input Sequence) to understand the context before writing the sentence in French (the Output Sequence).

AI does the exact same thing. But instead of translating English to French, it translates your Question into an Answer.
Representing Text as a Sequence

To process your question, the AI must first convert your text into a sequence of numbers. In the simplest method, the AI treats every word (and punctuation mark!) separately.

    Input Text: "How are you today?"

    The AI sees an ordered sequence of 5 items: ["How", "are", "you", "today", "?"]

Notice that the question mark ? is its own separate item. This list is the sequence that the AI will turn into numbers. We'll explore how this is done next.
1-ii. Representing Text Using Tokens 📘

Under the hood, chatbots don't work directly using words. Instead, they convert the words in a sequence of text into numerical values.

To do this, they build a Vocabulary (think of it like a dictionary or a menu).

    They list every unique word they know.
    They assign each unique word a specific ID number. These are known as tokens.

Whenever a word appears in a sentence (even if it appears multiple times), it is always replaced by that same specific token ID number.

Let's try generating a vocabulary of tokens using the text sequence from the previous section.
# --- TEXT INPUT ---

# We start with a normal Text String

input_text = "How are you today?"

print(f"Input text (String):       '{input_text}'")



# --- CLEANING ---

# Computers are very literal. To a computer, "today" and "today?" look like different words.
# We need to separate the '?' so the computer treats it as its own token.
# We do this by replacing every "?" with " ?" (adding a space before it).

text_processed = input_text.replace("?", " ?")

print(f"Punctuation fix (String):  '{text_processed}'")



# --- SEPARATE INTO PARTS ---

# Now we can safely chop the sentence wherever there is a space.

text_sequence = text_processed.split()

print(f"Text sequence (List):      {text_sequence}")



# --- BUILDING VOCABULARY ---

# We look for unique words (set) and number them (enumerate)
# to build a vocabulary that matches words to unique token IDs

vocabulary = {}
for i, word in enumerate(set(text_sequence)):
  vocabulary[word] = i

print(f"Vocabulary (Dict):         {vocabulary}")



# --- ENCODING ---

# Finally, we swap the words in our list for their matching token IDs

input_tokens = []
for word in text_sequence:
  input_tokens.append(vocabulary[word])

print(f"Input tokens (List):       {input_tokens}")

Input text (String):       'How are you today?'
Punctuation fix (String):  'How are you today ?'
Text sequence (List):      ['How', 'are', 'you', 'today', '?']
Vocabulary (Dict):         {'are': 0, 'you': 1, 'today': 2, '?': 3, 'How': 4}
Input tokens (List):       [4, 0, 1, 2, 3]

Going Backwards (Text Decoding) 🔄

The process above is called Encoding. We turned a human sentence into a list of numbers: [1, 2, 4, 3, 0].

But when an AI finishes thinking, it needs to talk back to you! To do this, it reverses the process (Decoding). We need a "Reverse Dictionary" where we can look up the ID Number to find the Word.
# --- INVERT THE VOCABULARY ---

# We need to swap the keys and values.
# Old Vocabulary:  Word -> Number
# New Vocabulary:  Number -> Word

vocabulary_inverted = {}
for word, number in vocabulary.items():
    vocabulary_inverted[number] = word

print(f"Reverse Vocabulary (Dict): {vocabulary_inverted}")



# --- DECODING ---

# Now we look up the token ID numbers to get the words back.

decoded_list = []
for number in input_tokens:
    decoded_list.append(vocabulary_inverted[number])

print(f"Decoded List (List):       {decoded_list}")

Reverse Vocabulary: {0: 'are', 1: 'you', 2: 'today', 3: '?', 4: 'How'}
Decoded List:       ['How', 'are', 'you', 'today', '?']

Section 1-ii. Summary

You have just built a complete data pipeline for processing text!

    Tokenization: You chopped a sentence into individual pieces (tokens).
    Vocabulary: You created a dictionary that gave every unique token a specific ID.
    Encoding: You converted the text into a list of numbers: [4, 1, 2, 0, 3].
    Decoding: You successfully converted those numbers back into the original text.
